```python

import pandas as pd

orders = pd.read_csv(DATA_DIR / "olist_orders_dataset.csv")

print("Dataset shape:", orders.shape)

print("\nOrder status counts:")
display(orders["order_status"].value_counts(dropna=False))

print("\nMissing values:")
display(orders.isna().sum().to_frame("Missing Values"))

print("\nTimestamp columns:")
display(orders[
    [
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
].head())
```

```python

# 1. Cancellation Rate
total_orders = len(orders)

cancelled_orders = (
    orders["order_status"].eq("canceled").sum()
)

cancellation_rate = (
    cancelled_orders / total_orders * 100
    if total_orders else 0
)

# 2. On-Time Delivery Rate
delivered = orders[
    orders["order_status"].eq("delivered")
].copy()

delivered["order_delivered_customer_date"] = pd.to_datetime(
    delivered["order_delivered_customer_date"],
    errors="coerce"
)

delivered["order_estimated_delivery_date"] = pd.to_datetime(
    delivered["order_estimated_delivery_date"],
    errors="coerce"
)

valid_delivery = delivered.dropna(
    subset=[
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
)

on_time_orders = (
    valid_delivery["order_delivered_customer_date"]
    <= valid_delivery["order_estimated_delivery_date"]
).sum()

on_time_rate = (
    on_time_orders / len(valid_delivery) * 100
    if len(valid_delivery) else 0
)

print("CANCELLATION RATE")
print(f"Cancelled orders: {cancelled_orders:,}")
print(f"Total orders: {total_orders:,}")
print(f"Cancellation rate: {cancellation_rate:.2f}%")

print("\nON-TIME DELIVERY RATE")
print(f"Eligible delivered orders: {len(valid_delivery):,}")
print(f"On-time orders: {on_time_orders:,}")
print(f"On-time delivery rate: {on_time_rate:.2f}%")
```

```python

# Convert timestamp columns
date_columns = [
    "order_purchase_timestamp",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")

# Use delivered orders for delivery KPIs
delivered = orders[
    orders["order_status"].eq("delivered")
].copy()

# KPI 3: Average delivery duration in days
delivered["delivery_duration_days"] = (
    delivered["order_delivered_customer_date"]
    - delivered["order_purchase_timestamp"]
).dt.total_seconds() / 86400

avg_duration = delivered["delivery_duration_days"].mean()

# KPI 4: Average delay compared with estimated date
delivered["delay_days"] = (
    delivered["order_delivered_customer_date"]
    - delivered["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

avg_delay = delivered["delay_days"].mean()

late_orders = delivered.loc[delivered["delay_days"] > 0]
avg_late_delay = late_orders["delay_days"].mean()

# KPI 5: Order processing time
# Purchase to carrier handoff; not warehouse-only processing
delivered["processing_days"] = (
    delivered["order_delivered_carrier_date"]
    - delivered["order_purchase_timestamp"]
).dt.total_seconds() / 86400

avg_processing = delivered["processing_days"].mean()

print("DELIVERY PERFORMANCE KPIs")
print(f"Average delivery duration: {avg_duration:.2f} days")
print(f"Average signed delivery delay: {avg_delay:.2f} days")
print(f"Average delay among late orders: {avg_late_delay:.2f} days")
print(f"Average purchase-to-carrier time: {avg_processing:.2f} days")
print(f"Delivered orders: {len(delivered):,}")
print(f"Late orders with valid dates: {len(late_orders):,}")
```

```python

reviews = pd.read_csv(DATA_DIR / "olist_order_reviews_dataset.csv")

# Keep one rating per order for this comparison
reviews = reviews[
    ["order_id", "review_score"]
].dropna().drop_duplicates()

# Merge reviews with delivery information
rating_data = delivered[
    [
        "order_id",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
].merge(reviews, on="order_id", how="inner")

# Classify delivery performance
rating_data["delivery_performance"] = np.where(
    rating_data["order_delivered_customer_date"]
    <= rating_data["order_estimated_delivery_date"],
    "On Time",
    "Late"
)

# Compare average ratings
rating_comparison = (
    rating_data.groupby("delivery_performance")
    .agg(
        Average_Rating=("review_score", "mean"),
        Orders_With_Reviews=("order_id", "nunique")
    )
    .reset_index()
)

display(rating_comparison)
```

```python

customers = pd.read_csv(DATA_DIR / "olist_customers_dataset.csv")

# Attach customer location to each order
regional_data = delivered.merge(
    customers[["customer_id", "customer_state"]],
    on="customer_id",
    how="left"
)

# Keep orders with valid delivery dates
regional_data = regional_data.dropna(
    subset=[
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
).copy()

# Calculate regional KPIs
regional_data["on_time"] = (
    regional_data["order_delivered_customer_date"]
    <= regional_data["order_estimated_delivery_date"]
)

regional_data["delivery_days"] = (
    regional_data["order_delivered_customer_date"]
    - regional_data["order_purchase_timestamp"]
).dt.total_seconds() / 86400

regional_data["delay_days"] = (
    regional_data["order_delivered_customer_date"]
    - regional_data["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

regional_summary = (
    regional_data.groupby("customer_state")
    .agg(
        Total_Orders=("order_id", "nunique"),
        On_Time_Rate=("on_time", "mean"),
        Avg_Delivery_Days=("delivery_days", "mean"),
        Avg_Delay_Days=("delay_days", "mean")
    )
    .reset_index()
)

regional_summary["On_Time_Rate"] *= 100

regional_summary = regional_summary.sort_values(
    "On_Time_Rate", ascending=False
)

display(regional_summary.head(10).round(2))
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

top_regions = regional_summary.head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_regions,
    x="customer_state",
    y="On_Time_Rate"
)

plt.title("Top 10 Customer States by On-Time Delivery Rate")
plt.xlabel("Customer State")
plt.ylabel("On-Time Delivery Rate (%)")
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Keep regions with sufficient order volume
eligible_regions = regional_summary[
    regional_summary["Total_Orders"] >= 100
].copy()

# Select the 10 lowest on-time rates
bottom_regions = eligible_regions.sort_values(
    "On_Time_Rate", ascending=True
).head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=bottom_regions,
    x="customer_state",
    y="On_Time_Rate"
)

plt.title("10 States with Lowest On-Time Delivery Rates")
plt.xlabel("Customer State")
plt.ylabel("On-Time Delivery Rate (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Create a master KPI summary

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Orders",
        "Cancelled Orders",
        "Cancellation Rate (%)",
        "Eligible Delivered Orders",
        "On-Time Orders",
        "On-Time Delivery Rate (%)",
        "Average Delivery Duration (Days)",
        "Average Signed Delay (Days)",
        "Average Delay Among Late Orders (Days)",
        "Average Purchase-to-Carrier Time (Days)"
    ],
    "Value": [
        len(orders),
        cancelled_orders,
        round(cancellation_rate, 2),
        len(valid_delivery),
        int(on_time_orders),
        round(on_time_rate, 2),
        round(avg_duration, 2),
        round(avg_delay, 2),
        round(avg_late_delay, 2),
        round(avg_processing, 2)
    ]
})

display(kpi_summary)
```

```python

from pathlib import Path

OUTPUT_DIR = Path("../data/processed")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

kpi_summary.to_csv(
    OUTPUT_DIR / "delivery_kpi_summary.csv",
    index=False
)

regional_summary.to_csv(
    OUTPUT_DIR / "regional_performance.csv",
    index=False
)

rating_comparison.to_csv(
    OUTPUT_DIR / "customer_rating_comparison.csv",
    index=False
)

print("Export completed.")
print("Saved files:")
for file in sorted(OUTPUT_DIR.glob("*.csv")):
    print(file.name)
```
